```python
import cv2
import matplotlib.pyplot as plt

image_path='/content/drive/MyDrive/image.jpg'
image=cv2.imread(image_path)
image_rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8,8))
plt.imshow(image_rgb)
plt.axis('off')
plt.title('Original Image')
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

input_image_path='/content/drive/MyDrive/image.jpg'
gray_conversion=cv2.imread(input_image_path,cv2.IMREAD_GRAYSCALE)

threshold=int(input())
print(f"Mean Intensity Threshold={threshold:.2f}")

binary=np.where(gray_conversion>=threshold,255,0).astype(np.uint8)

plt.figure(figsize=(8,8))
plt.imshow(binary)
plt.axis('off')
plt.title('Binary Image')
plt.show()

plt.figure(figsize=(8,8))
plt.imshow(gray_conversion)
plt.axis('off')
plt.title('Gray Conversion')
plt.show()

height, width = gray_conversion.shape

pixels = gray_conversion # Define 'pixels' as the grayscale image data

total=0.0
for y in range(height):
  for x in range(width):
    total+=pixels[y,x] # Corrected indexing from pixels[x,y] to pixels[y,x]
mean_threshold=total/(width*height)
print(total)
print("Mean Threshold =", mean_threshold)
```
