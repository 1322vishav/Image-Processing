```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image_path='/content/drive/MyDrive/image.jpg'
image=cv2.imread(image_path)
image_rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8,8))
plt.imshow(image_rgb)
plt.axis('off')
plt.title('Original Image')
plt.show()
```

```python
# Case 1: Grayscale by mean average of R, G, B planes

height, width, channels = image_rgb.shape

R = image_rgb[:,:,0]
G = image_rgb[:,:,1]
B = image_rgb[:,:,2]

mean_gray = np.zeros((height, width), dtype=np.uint8)

for y in range(height):
  for x in range(width):
    mean_gray[y,x] = (int(R[y,x]) + int(G[y,x]) + int(B[y,x])) / 3

plt.figure(figsize=(8,8))
plt.imshow(mean_gray, cmap='gray')
plt.axis('off')
plt.title('Grayscale - Mean Average')
plt.show()
```

```python
# Case 2: Grayscale by user-defined weightage of R, G, B planes

wR = float(input("Enter weightage for R: "))
wG = float(input("Enter weightage for G: "))
wB = float(input("Enter weightage for B: "))

print(f"Sum of weightages = {wR + wG + wB}")

weighted_gray = np.zeros((height, width), dtype=np.uint8)

for y in range(height):
  for x in range(width):
    weighted_gray[y,x] = (wR*int(R[y,x]) + wG*int(G[y,x]) + wB*int(B[y,x]))

plt.figure(figsize=(8,8))
plt.imshow(weighted_gray, cmap='gray')
plt.axis('off')
plt.title(f'Grayscale - Weighted (R={wR}, G={wG}, B={wB})')
plt.show()
```
