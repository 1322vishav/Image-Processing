```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image_path='/content/drive/MyDrive/image.jpg'
image=cv2.imread(image_path)
image_rgb=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)

threshold=int(input("Enter threshold for binary image: "))
binary=np.where(gray>=threshold,255,0).astype(np.uint8)

plt.figure(figsize=(8,8))
plt.imshow(image_rgb)
plt.axis('off')
plt.title('Original Image')
plt.show()
```

## Q1: Border / Padding

```python
# Border on grayscale image

border_width=int(input("Enter border width (pixels): "))
border_color_gray=int(input("Enter border color (0-255): "))

h, w = gray.shape
gray_bordered = np.full((h + 2*border_width, w + 2*border_width), border_color_gray, dtype=np.uint8)
gray_bordered[border_width:border_width+h, border_width:border_width+w] = gray

plt.figure(figsize=(8,8))
plt.imshow(gray_bordered, cmap='gray')
plt.axis('off')
plt.title('Grayscale Image with Border')
plt.show()
```

```python
# Border on binary image

border_width_bin=int(input("Enter border width (pixels): "))
border_color_bin=int(input("Enter border color (0 or 255): "))

h, w = binary.shape
binary_bordered = np.full((h + 2*border_width_bin, w + 2*border_width_bin), border_color_bin, dtype=np.uint8)
binary_bordered[border_width_bin:border_width_bin+h, border_width_bin:border_width_bin+w] = binary

plt.figure(figsize=(8,8))
plt.imshow(binary_bordered, cmap='gray')
plt.axis('off')
plt.title('Binary Image with Border')
plt.show()
```

## Q2: Image Complement

```python
# Grayscale complement

gray_complement = 255 - gray

plt.figure(figsize=(8,8))
plt.imshow(gray_complement, cmap='gray')
plt.axis('off')
plt.title('Grayscale Complement')
plt.show()
```

```python
# Binary complement

binary_complement = 255 - binary

plt.figure(figsize=(8,8))
plt.imshow(binary_complement, cmap='gray')
plt.axis('off')
plt.title('Binary Complement')
plt.show()
```
